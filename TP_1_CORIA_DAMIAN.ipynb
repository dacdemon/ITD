```python
# IMPORTACION DE LIBRERIAS
import random
```

```python

# GENERACION DE PRECIOS

todos_los_precios = []

for i in range(100):
  precio = random.randint(10, 1000)
  todos_los_precios.append(precio)

# FILTRADO DE PRECIOS PREMIUM

precios_premium = []

for precio in todos_los_precios:
  if precio > 300:
    precios_premium.append(precio)

# CALCULO Y APLICACION DEL 21% DE IVA

precios_con_iva = []

for precio in precios_premium:
  precio_iva = precio * 1.21
  precios_con_iva.append(precio_iva)

# REPORTE FINAL

cantidad_productos = len(precios_premium)
total_facturacion = sum(precios_con_iva)
total_facturacion_formateado = round(total_facturacion, 2)

print("CANTIDAD DE PRODUCTOS QUE SUPERARON LOS $300:", cantidad_productos)
print("TOTAL DE FACTURACION CON IVA INCLUIDO: $", total_facturacion_formateado)
```

# CONCLUSION
En esta practica se pudo generar una lista de 100 precios al azar y se realizo un proceso de filtrado para seleccionar solo los productos con un valor superior a los $300.
Tambien se aplico un 21% de iva a los precios seleccionados anteriormente.
Finalmente se calculo la cantidad de precios filtrados y el total de facturacion con iva incluido.
Los resultados van a ser diferentes cada vez que se ejecute ya que los mismos se generan al azar.
